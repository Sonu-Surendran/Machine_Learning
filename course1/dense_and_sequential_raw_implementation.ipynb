```python
import numpy as np
```

```python
def dense(data, w, b):
    n = w.shape[1]
    res = np.zeros(n)

    for j in range(n):
        w1 = w[:, j]
        pred = data @ w1 + b[j]
        a = sigmoid(pred)
        res[j] = a
    return res
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def sequential(x_train):
    w1 = np.array([[10, -9, 12], [2, 20, -2]])
    b1 = np.array([6, 4, 9])
    a1 = dense(x_train, w1, b1)
    w2 = np.array([[-1, 2], [-12, -8], [1, 23]])
    b2 = np.array([12, 32])
    a2 = dense(a1, w2, b2)
    fx = a2
    return fx
```

```python
a = np.array([10, 2])
```

```python
sequential(a)
```
