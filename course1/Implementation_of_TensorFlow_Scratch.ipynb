```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([-3, -4])
```

Implementation of first neuron - Layer 1

```python
def sigmoid(z):
    resu = 1/(1+np.exp(-z))
    return resu
```

```python
w1_1 = np.array([-10, 3])
b1_1 = 34
pred1_1 = x_train@w1_1 + b1_1
a1_1 = sigmoid(pred1_1)
```

Similar implementation of other neuron in layer 1

```python
w2_1 = np.array([3, 9])
b2_1 = 2
pred2_1 = x_train@w2_1 + b2_1
a2_1 = sigmoid(pred2_1)

w3_1 = np.array([45, 9])
b3_1 = 67
pred3_1 = x_train@w3_1 + b3_1
a3_1 = sigmoid(pred3_1)
```

We have three neuron in the first layer for which the combained output of all the neuron is stored in a1

```python
a1 = np.array([a1_1, a2_1, a3_1])
```

Now lets pass this new output from layer one to two layer as a inputs

```python
w1_2 = np.array([-4, 43, 7])
b1_2 = -23

pred1_2 = a1@w1_2 + b1_2
a1_2 = sigmoid(pred1_2)

w2_2 = np.array([34, -56, -3])
b2_2 = 3
pred2_2 = w2_2 @ a1 + b2_2
a2_2 = sigmoid(pred2_2)

a2 = np.array([a1_2, a2_2])
```

Finally passing the output of layer2 to output layer

```python
w1_3 = np.array([7, -23])
b1_3 = 34
pred1_3 = w1_3 @ a2 + b1_3
a1_3 = sigmoid(pred1_3)
```

```python
fx_wb = a1_3

if fx_wb >= 0.5:
    print(1)
else:
    print(0)
```
