```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/archive/cv_server_data.csv")
```

```python
plt.scatter(df.iloc[:,0], df.iloc[:,1], label="data plot")
plt.show()
```

```python
data_matrix = np.array(df)
```

```python
data_matrix.shape
```

```python
def compute_coefi(X):

    mew = np.mean(X, axis=0)
    cor = np.var(X, axis=0)

    return mew, cor
```

```python
mew, var = compute_coefi(data_matrix)
```

```python
def compute_prob(x, mew, var):

    left_hand = 1/(np.sqrt((2*np.pi*var)))
    right_hand = np.exp(((x - mew)**2)/(2*var))
    prob = left_hand*right_hand

    return np.prod(prob, axis=1)
```

```python
pred = compute_prob(data_matrix, mew, var)
```

```python
pred.shape
```

```python
pred = (pred > 3e2).astype(int)
    
```
