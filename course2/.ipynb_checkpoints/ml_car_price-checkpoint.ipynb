```python
def linear_regression(xi, wj, w0):
    n = len(xi)
    pred = w0
    
    for j in range(n):
        pred += xi[j] * wj[j]
    
    return pred
```

## 2.6 Linear Regression vector form

```python
def dot_(xi, w):
    res = 0

    m = len(xi)

    for i in range(m):
        res = res + xi[i] * w[i]
    return res
```

```python
new_w = [w0] + w
```

```python
new_w
```

```python
def l_r(xi, w):
    return w0 + dot_(xi,w)
```

```python
l_r(xi,w)
```

## 2.7 Training a linear regression model

```python
X = [
    [148, 24, 1385],
    [132, 25, 2031],
    [453, 11, 86],
    [158, 24, 185],
    [172, 25, 201],
    [413, 11, 86],
    [38,  54, 185],
    [142, 25, 431],
    [453, 31, 86],
]

X = np.array(X)
X
```

```python
ones = np.ones(X.shape[0])
X = np.column_stack((ones, X))
```

```python
X
```

```python
y = [10000, 20000, 15000, 20050, 10000, 20000, 15000, 25000, 12000]
```

```python
def regression(X, Y):

    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    reg = np.eye(X.shape[1])
    reg = 0.001 * reg
    
    xtx = (X.T).dot(X) + reg
    xtx_1 = np.linalg.inv(xtx)
    w_weight = xtx_1.dot(X.T).dot(Y)
    return w_weight[0], w_weight[1:]
```

```python
regression(X, y)
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    return w_full[0], w_full[1:]
```

```python
train_linear_regression(X,y)
```

## 2.9 RMSE

```python
def rmse(y_pred, y):
    se = (y_pred - y)**2
    mse = np.mean(se)
    rmse = np.sqrt(mse)
    return rmse
```

```python
rmse(pred, y_train)
```

## 2.10 Validating Model

```python
def data_preparation(X, features):
    data = X.copy()

    #create age feature for the car
    mx_year = np.max(data["year"])
    year_val = data["year"].values
    data["age"]= mx_year - year_val
    features = list(features) + ["age"]

    #numbers of door one hot encoding
    for i in [2,3,4]:
        data["num_door_%s"% i] = (data.number_of_doors == i).astype(int)
        features.append("num_door_%s" % i)

    data = data[features]
    data = data.fillna(0)

    data = data.values
    return data
```

```python
value_col =['engine_hp', 'engine_cylinders', 'highway_mpg',
        'city_mpg', 'popularity']
x_train = data_preparation(df_train, value_col)
```

```python
x_train
```

```python
x_features = df_train.dtypes[(df_train.dtypes == "int64") | (df_train.dtypes == "float64")].index
```

```python
x_features
```

```python
x_train = data_preparation(df_train, x_features)
```

```python
w0, w = regression(x_train, y_train)
```

```python
pred = w0+x_train.dot(w)
```
