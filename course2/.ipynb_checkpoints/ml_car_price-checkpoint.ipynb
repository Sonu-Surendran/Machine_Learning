```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/cardataset/data.csv")
```

```python
df.head()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
df.head()
```

```python
for col in df.columns:
    print(col, df[col].nunique())
    print(df[col].unique()[:5])
    print()
```

```python
df.dtypes
```

```python
df.dtypes == "object"
```

```python
str_col = df.dtypes[df.dtypes=="object"]
```

```python
str_col
```

```python
for column in str_col.index:
    df[column] = df[column].str.lower().str.replace(" ", "_")
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df.msrp, bins=40)
plt.xlabel("Price")
plt.ylabel("Frequency of car")
plt.title("Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(4,6))
sns.histplot(df.msrp[df.msrp <= 200000], bins=40)
plt.xlabel("price")
plt.ylabel("Frequency of cars")
plt.title("Pricing Distribution of car")
plt.show()
```

```python
df.msrp = np.log(df.msrp)
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df.msrp, bins=40)
plt.xlabel("price")
plt.ylabel("frequency of car")
plt.title("Distribution of car pricing")
plt.show()
```
