```python
import requests

# 1. The URL where your FastAPI backend is running
url = "http://127.0.0.1:8000/predict"

# 2. Path to the image file you want to test
image_path = "./data/images/01f52917-7e92-4c47-8979-1ab9a41c96a8.jpg" 

# 3. Open the file in binary read mode ("rb")
with open(image_path, "rb") as img_file:
    # The key name "file" must exactly match the variable name 
    # in your FastAPI endpoint: `predict(file: UploadFile = File(...))`
    files = {"file": img_file}
    
    try:
        # Send the POST request
        response = requests.post(url, files=files)
        
        # Check if the request was successful
        if response.status_code == 200:
            result = response.json()
            print("Successfully received prediction:")
            print(result)
        else:
            print(f"Failed with status code: {response.status_code}")
            print("Response:", response.text)
            
    except requests.exceptions.ConnectionError:
        print("Could not connect to the backend. Is your FastAPI server running?")
```
